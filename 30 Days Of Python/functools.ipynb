```python
from functools import (
    cache, cached_property, cmp_to_key, lru_cache, total_ordering,
    partial, partialmethod, reduce, singledispatch, singledispatchmethod,
    update_wrapper, wraps,
)
import operator
```

```python
@cache
def slow_square(n):
    print(f"computing {n}")
    return n * n
```

```python
slow_square(4)   # prints "computing 4" → 16
# slow_square(4)   # → 16   (no print — served from cache)
```

```python
def greet(greeting,name):
    return f"{greeting}, {name}"

hello = partial(greet, "Hello")
hello("Arun")
```

```python
def log_calls(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper
@log_calls
def greet(name):
    "Say, Hi"
    return f"Hi, {name}"
```

```python
greet("Arun")
```

```python
@cache
def fact(n):
    if n <= 1:
        return 1
    return n * fact(n - 1)
```

```python
print(fact(10))
```

```python
fact(5)
```

```python
fact.cache_info()
```

```python
def log_calls2(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling of {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_calls2
def say_my_name(name):
    print(f"Hello, You are {name}")
```

```python
say_my_name("Aadhi")
```

```python
print(say_my_name.__name__)
```

```python
@lru_cache(maxsize=3)
def squares(n):
    print(f"Computing sqaure({n})")
    return n**n
```

```python
squares(5)
```

```python
squares.cache_info()
```
