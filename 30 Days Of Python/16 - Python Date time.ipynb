```python
from datetime import datetime
now = datetime.now()
print(now)
day = now.day
month = now.month
year = now.year
hour = now.hour
minute = now.minute
second = now.second
timestamp = now.timestamp()

print(day, month, year, minute, second, hour)
print(f"Timestamp: {timestamp}")
print(f"{day}/{month}/{year}, {hour}:{minute}")
```

```python
from datetime import datetime

now = datetime.now()
t =  now.strftime("%H:%M:%S")

print(f"Time: ",t)

time_one = now.strftime("%d/%m/%Y %H:%M:%S")
print(f"Time One: ", time_one)
```
