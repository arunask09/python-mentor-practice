```python
# Function Application and Mapping
```

```python
frame = pd.DataFrame(np.random.standard_normal((4, 3)), columns=list("bde"),index=["Utah", "Ohio", "Texas", "Oregon"])
```

```python
frame
```

```python
np.abs(frame)
```

```python
def f1(x):
    return x.max() - x.min()

frame.apply(f1)
```

```python
frame.apply(f1, axis="columns")
```

```python
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])

frame.apply(f2)
```

```python
frame
```

```python
def my_format(x):
    return f"{x:.2f}"

frame.map(my_format)
```

```python
import timeit
n = 1_000_000
lst, st = list(range(n)), set(range(n))
print(timeit.timeit(lambda: (n - 1) in lst, number=100))  # scans all n elements
print(timeit.timeit(lambda: (n - 1) in st,  number=100))  # jumps straight there
```
