```python
# nums = [1, -3, 4, -7, 9, 0, 2] —
# build a list of squares of only the positive numbers, using walrus to compute each square once (inside the if) 
# and reuse it as the output value.
```

```python
nums = [1, -3, 4, -7, 9, 0, 2]
squares = [y for num in nums if num > 0 and (y := num**2)]
squares
```

```python
# ops = {
#     "add": lambda a, b: a + b,
#     "sub": lambda a, b: a - b,
#     "mul": lambda a, b: a * b,
# }

# Then write a function calculate(op_name, a, b) that looks up the right lambda in ops by name and calls it with a and b. 
# For example, calculate("mul", 4, 5) should return 20.
```

```python
ops = {
    "add": lambda a, b: a + b,
    "sub": lambda a, b: a - b,
    "mul": lambda a, b: a * b,
}
```

```python
def calculate(op_name, a, b):
    return ops[op_name](a, b)

calculate("mul", 4, 5)
```

```python
print(calculate('add',10,20))
```

```python
# M2 (filter) — Given 
# books = [{"title": "It", "pages": 1138}, {"title": "Carrie", "pages": 199}, {"title": "The Shining", "pages": 447}], 
# use filter to get a list of just the book dicts with more than 300 pages.
```

```python
books = [{"title": "It", "pages": 1138}, {"title": "Carrie", "pages": 199}, 
         {"title": "The Shining", "pages": 447}]

book_list = list(map(lambda book: book["title"], filter(lambda book: book["pages"] > 300, books)))

book_list
```
