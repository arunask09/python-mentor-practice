```python
# TWO SUM
```

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        seen = {}
        for i, num in enumerate(nums):
            diff = target - num
            if diff in seen:
                return [seen[diff], i]
            seen[num] = i
        return []
```

```python
sol = Solution()
print(sol.twoSum([4,2,1,5,6], 9))
```
