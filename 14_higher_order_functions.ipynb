```python
# B — Bold
def bold_deco(function):
    def wrapper(*args, **kwargs):
        result = function(*args, **kwargs)
        return f"\033[1m{result}\033[0m"
    return wrapper

# I — Italic
def italic_deco(function):
    def wrapper(*args, **kwargs):
        result = function(*args, **kwargs)
        return f"\033[3m{result}\033[0m"
    return wrapper

# A — Argument-taking decorator (this is the new piece)
def repeat(times):
    def actual_decorator(function):
        def wrapper(*args, **kwargs):
            results = []
            for _ in range(times):
                results.append(function(*args, **kwargs))
            return "\n".join(results)
        return wrapper
    return actual_decorator


@bold_deco
@italic_deco
@repeat(times=3)
def greeting(name):
    return f"Hello, {name}"


def main():
    user_name = input("Enter the user name: ")
    print(greeting(user_name))

main()
```

```python
# Write a decorator called validate_positive that checks the arguments passed into a function 
# (not the return value) and raises a ValueError if any numeric argument is negative, before letting the function run.
```

```python
def validate_positive(function):
    def wrapper(*args, **kwargs):
        for number in args:
            if isinstance(number,(int, float)) and number < 0:
                raise ValueError (f"Given number is negative, got {number}")
        return function(*args, **kwargs)
    return wrapper
    
@validate_positive
def give_number(num):
    return num

def main():
    user_value = float(input("Enter the number: "))
    print(give_number(user_value))

main()
```
